```python
import pandas as pd
```

```python
df = pd.read_csv("Naive-Bayes-Classification-Data.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
x = df.drop("diabetes",axis=1)
y = df["diabetes"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.20, random_state = 42
)
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
#x_test_scaled
```

```python
model = GaussianNB()
```

```python
model
```

```python
model.fit(x_train_scaled,y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred
```

```python
comparison = pd.DataFrame({
    "Actual": y_test,
    "Prediction": y_pred
})
```

```python
comparison.head(10)
```

```python
accuracy = accuracy_score(y_test,y_pred)*100

print(f"accuracy: {accuracy:.2f}%")
```

```python
print("Classification Report :\n", classification_report(y_test,y_pred))
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
print(cm)
```

```python
for i in range(45):
    x_train, x_test, y_train, y_test = train_test_split(
        x,y,test_size = 0.2,random_state = i
    )
    model = GaussianNB()
    model.fit(x_train,y_train)
    print(i, accuracy_score(y_test,model.predict(x_test)))
```

```python
from  sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(GaussianNB(),x,y,cv = 5)
print(scores)
```

```python
print(scores.mean()*100)
```

## Model Evaluation

The Diabetes Classification model achieved an accuracy of **92.96%** on the test dataset.

The classification report shows that both classes performed well, with precision, recall, and F1-score around **92%–93%**.

The confusion matrix shows:
- **86** instances of Class 0 were correctly classified.
- **99** instances of Class 1 were correctly classified.
- **7** Class 0 instances were incorrectly classified as Class 1.
- **7** Class 1 instances were incorrectly classified as Class 0.

The model achieved an average **cross-validation score of 93.27%**, indicating good and consistent performance across different data splits.

### Conclusion

Overall, the model performed well on the Diabetes dataset, achieving approximately **93% accuracy** with balanced precision, recall, and F1-score for both classes.
